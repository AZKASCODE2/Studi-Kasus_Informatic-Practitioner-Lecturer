A. Studi Kasus: Dashboard Monitoring Keterisian Unit Perawatan Intensif (ICU) & Prioritas Triase IGD Rumah Sakit
Latar Belakang & Masalah: Tenaga medis dan direktur rumah sakit dalam situasi tanggap darurat kerap disajikan dashboard dengan visual clutter (grafik 3D yang berlebihan, warna bertabrakan, dan ketiadaan garis batas kritis). Kondisi ini meningkatkan beban kognitif dokter saat harus memutuskan pengalihan pasien darurat secara cepat.
Solusi: Merancang dashboard analitik berbasis pedoman HCI:
F-Pattern Layout: Menaruh metrik KPI paling kritis di kuadran kiri atas yang pertama kali dipindai mata pengguna.
Visual Thresholding: Memberikan garis penanda batas darurat (misal keterisian 85%) dengan zona arsiran merah transparan agar status darurat langsung terbaca tanpa perlu membaca angka detail.
Direct Data Labeling: Menyematkan angka langsung pada batang horizontal triase guna menghilangkan proses mental mencocokkan sumbu X secara berulang.

```python
import matplotlib
matplotlib.use('Agg') # Render tanpa display GUI eksternal
import matplotlib.pyplot as plt
import numpy as np

def studi_kasus_5_hci_visualization():
    np.random.seed(42)
    jam = [f"{h:02d}:00" for h in range(24)]
    # Simulasi keterisian ICU (%) selama 24 jam terakhir
    icu_occ = np.clip(np.cumsum(np.random.normal(0.5, 2.5, size=24)) + 70, 40, 96)

    # Kategori Pasien Triase IGD
    kategori_triase = ['Resusitasi (P1)', 'Emergensi (P2)', 'Urgensi (P3)', 'Non-Urgensi (P4)']
    jumlah_pasien = [4, 12, 28, 15]
    # Warna semantik aman & jelas (Merah = Kritis, Biru Muda = Aman)
    warna_triase = ['#d73027', '#fc8d59', '#fee090', '#91bfdb']

    fig = plt.figure(figsize=(11, 6), constrained_layout=True)
    gs = fig.add_gridspec(2, 2, width_ratios=[1.2, 1], height_ratios=[1, 2])

    # 1. KPI Card Utama (Prinsip HCI: F-Pattern di Kiri Atas)
    ax_kpi = fig.add_subplot(gs[0, 0])
    ax_kpi.axis('off')
    current_occ = icu_occ[-1]
    status_str = "WASPADA (ALOKASIKAN RUANGAN)" if current_occ > 85 else "KAPASITAS AMAN"
    warna_status = "#b2182b" if current_occ > 85 else "#2166ac"

    ax_kpi.text(0.05, 0.75, "KETERISIAN ICU SAAT INI", fontsize=10, fontweight='bold', color='#555555')
    ax_kpi.text(0.05, 0.30, f"{current_occ:.1f}%", fontsize=26, fontweight='bold', color=warna_status)
    ax_kpi.text(0.05, 0.05, f"Status: {status_str}", fontsize=9, fontweight='bold', color=warna_status)

    # 2. KPI Waktu Tunggu Triase Rata-rata
    ax_kpi2 = fig.add_subplot(gs[0, 1])
    ax_kpi2.axis('off')
    ax_kpi2.text(0.05, 0.75, "RATA-RATA RESPON TRIASE IGD", fontsize=10, fontweight='bold', color='#555555')
    ax_kpi2.text(0.05, 0.30, "14.2 Menit", fontsize=26, fontweight='bold', color='#2b83ba')
    ax_kpi2.text(0.05, 0.05, "Target Standar Pelayanan Minimal: < 30 Menit", fontsize=9, color='#2c7bb6')

    # 3. Grafik Tren dengan Ambang Batas Kritis (Thresholding)
    ax_trend = fig.add_subplot(gs[1, 0])
    ax_trend.plot(jam, icu_occ, color='#2b83ba', linewidth=2.2, label='Okupansi')
    ax_trend.axhline(85, color='#d73027', linestyle='--', linewidth=1.5, label='Batas Kritis (85%)')
    ax_trend.fill_between(range(24), 85, 100, color='#d73027', alpha=0.12)
    ax_trend.set_title("Tren Okupansi ICU 24 Jam Terakhir", fontsize=10, fontweight='bold')
    ax_trend.set_ylabel("Keterisian (%)")
    ax_trend.set_xticks(range(0, 24, 4))
    ax_trend.grid(axis='y', linestyle=':', alpha=0.6)
    ax_trend.legend(loc='lower left', fontsize=8)

    # 4. Bar Horizontal dengan Direct Labeling (Beban Kognitif Rendah)
    ax_bar = fig.add_subplot(gs[1, 1])
    bars = ax_bar.barh(kategori_triase, jumlah_pasien, color=warna_triase, height=0.6)
    ax_bar.set_title("Antrean Pasien Berdasarkan Prioritas Triase", fontsize=10, fontweight='bold')
    ax_bar.set_xlabel("Jumlah Pasien")
    for b in bars:
        w = b.get_width()
        ax_bar.text(w + 0.6, b.get_y() + b.get_height()/2, f"{int(w)} pasien",
                    va='center', fontsize=9, fontweight='bold')
    ax_bar.set_xlim(0, max(jumlah_pasien) + 7)
    ax_bar.spines['top'].set_visible(False)
    ax_bar.spines['right'].set_visible(False)

    plt.savefig('dashboard_hci_icu.png', dpi=150)
    plt.close()
    print("=== STUDI KASUS 5: DESAIN DASHBOARD BERORIENTASI HCI ===")
    print("Dashboard berhasil digenerate dan disimpan ke 'dashboard_hci_icu.png'")
    print("- Prinsip F-Pattern: Ringkasan metrik kritis langsung terbaca di kuadran kiri atas.")
    print("- Prinsip Preeattentive Processing: Garis batas merah 85% memicu atensi visual instan.")
    print("- Prinsip Direct Labeling: Tidak membebani pengguna dengan pencocokan sumbu secara manual.")

if __name__ == '__main__':
    studi_kasus_5_hci_visualization()
```
